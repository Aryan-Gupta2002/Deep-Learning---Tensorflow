#Recurrent Neural Network.
---

###Sequential Odd numbers dataset
----

```python
import numpy as np

seq_len = 4

def create_dataset(series, seq_len=4):
    X, y = [], []

    for i in range(len(series) - seq_len):
        X.append(series[i:i+seq_len])
        y.append(series[i+seq_len])

    return np.array(X), np.array(y)

odd = np.arange(1, 100, 2)
X_odd, y_odd = create_dataset(odd, seq_len)
X_odd,y_odd

sequence_length=seq_len
max_val = np.max(odd)

X_odd = X_odd.reshape((X_odd.shape[0], X_odd.shape[1], 1))
X_odd = X_odd / max_val
y_odd = y_odd / max_val
```

###Tensorflow Implementation of RNN
---

```python
import tensorflow as tf
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.SimpleRNN(16, activation="tanh", input_shape=(sequence_length, 1)),
    layers.Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

model.fit(X_odd, y_odd, epochs=100)
```

Tensorflow Output

```python
test_seq = np.array([[21, 23, 25, 27]])
test_seq = test_seq.reshape((1, seq_len, 1))
test_seq = test_seq / max_val

pred = model.predict(test_seq)

print("Predicted:", pred[0][0] * max_val)
```

###Pytorch Implementation of RNN
----

```python
import torch
import torch.nn as nn
import torch.optim as optim

X_oddT = torch.tensor(X_odd, dtype=torch.float32)
y_oddT = torch.tensor(y_odd, dtype=torch.float32).view(-1,1)


class RNNModel(nn.Module):
    def __init__(self, input_size, hidden_size):
        super(RNNModel, self).__init__()
        self.rnn = nn.RNN(input_size, hidden_size, batch_first=True, nonlinearity='tanh')
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.rnn(x)
        out = out[:, -1, :]
        out = self.fc(out)
        return out

model = RNNModel(input_size=1, hidden_size=16)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# Example training loop
for epoch in range(100):
    optimizer.zero_grad()
    outputs = model(X_oddT)
    loss = criterion(outputs, y_oddT)
    loss.backward()
    optimizer.step()

    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss: {loss.item()}")
```

Pytorch Output

```python
model.eval()

test_seq = np.array([[21,23,25,27]])
test_seq = test_seq.reshape((1, seq_len, 1))
test_seq = test_seq / max_val

test_seq = torch.tensor(test_seq, dtype=torch.float32)

pred = model(test_seq).item()

print("Predicted:", pred * max_val)
```

##Scenerio 2 - 2.	Add additional sequence types and observe prediction behavior

```python
def prepare_sequence(series, seq_len):
    X, y = create_dataset(series, seq_len)

    max_val = np.max(series)

    X = X.reshape((X.shape[0], X.shape[1], 1))
    X = X / max_val
    y = y / max_val

    return X, y, max_val
mult3 = np.arange(3, 300, 3)
X_mult3, y_mult3, max_mult3 = prepare_sequence(mult3, seq_len)

model3 = models.Sequential([
    layers.SimpleRNN(16, activation="tanh", input_shape=(sequence_length, 1)),
    layers.Dense(1)
])

model3.compile(
    optimizer="adam",
    loss="mse"
)

model3.fit(X_mult3, y_mult3, epochs=100)
test = mult3[:seq_len]

test_norm = test / max_mult3
test_norm = test_norm.reshape(1, seq_len, 1)

pred = model3.predict(test_norm)

print("Predicted:", pred[0][0] * max_mult3)
```
